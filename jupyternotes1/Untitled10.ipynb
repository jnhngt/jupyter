```python
from pyecharts.charts import Bar
from pyecharts import options as opts
import pandas as pd
```

```python
df = pd.read_excel("水资源数据.xlsx")
```

```python
zong_data = df.iloc[0,1:]
zong_data
```

```python
dibiao_data = df.iloc[1,1:]
dibiao_data
```

```python
dixia_data = df.iloc[2,1:]
dixia_data
```

```python
year = zong_data.index.to_list()[::-1]
year
```

```python
water_zong = zong_data.values.tolist()[::-1]
water_dixia = dixia_data.values.tolist()[::-1]
water_dibiao = dibiao_data.values.tolist()[::-1]
```

```python
waterBar = (
    Bar()
    .add_xaxis(year)
    .add_yaxis("水资源总量",water_zong,
               markpoint_opts = opts.MarkPointOpts(data =[
                   opts.MarkPointItem(type_="max",name="最大值"),
                   opts.MarkPointItem(type_="min",name="最小值"),
                   opts.MarkPointItem(type_="average",name="平均值")
                                                         ]
                                                  )    
              )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="水资源总量变化折线图",subtitle="2016年-2024年"),
        yaxis_opts=opts.AxisOpts(name="总量(亿立方米)"),
        xaxis_opts=opts.AxisOpts(name="年份")
    )
)
waterBar.render("水资源总量变化柱状图.html")
```

```python
waterBar = (
    Bar()
    .add_xaxis(year)
    .add_yaxis("水资源总量",water_zong,
               markpoint_opts = opts.MarkPointOpts(data =[
                   opts.MarkPointItem(type_="max",name="最大值"),
                   opts.MarkPointItem(type_="min",name="最小值"),
                   opts.MarkPointItem(type_="average",name="平均值")
                                                         ]
                                                  )    
              )
    .add_yaxis("地下水资源总量",water_dixia)
    .add_yaxis("地表水资源总量",water_dibiao)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="水资源总量变化折线图",subtitle="2016年-2024年"),
        yaxis_opts=opts.AxisOpts(name="总量(亿立方米)"),
        xaxis_opts=opts.AxisOpts(name="年份")
    )
   
)
waterBar.render("水资源总量变化柱状图2.html")
```

```python
waterBar = (
    Bar()
    .add_xaxis(year)
    .add_yaxis("水资源总量",water_zong,stack="stack1",
               markpoint_opts = opts.MarkPointOpts(data =[
                   opts.MarkPointItem(type_="max",name="最大值"),
                   opts.MarkPointItem(type_="min",name="最小值"),
                   opts.MarkPointItem(type_="average",name="平均值")
                                                         ]
                                                  )    
              )
    .add_yaxis("地下水资源总量",water_dixia,stack="stack1",)
    .add_yaxis("地表水资源总量",water_dibiao,stack="stack1",)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="水资源总量变化折线图",subtitle="2016年-2024年"),
        yaxis_opts=opts.AxisOpts(name="总量(亿立方米)"),
        xaxis_opts=opts.AxisOpts(name="年份")
    )
   
)
waterBar.render("水资源总量变化柱状图3.html")
```

```python
waterBar = (
    Bar()
    .add_xaxis(year)
    .add_yaxis("水资源总量",water_zong,stack="stack1",
               markpoint_opts = opts.MarkPointOpts(data =[
                   opts.MarkPointItem(type_="max",name="最大值"),
                   opts.MarkPointItem(type_="min",name="最小值"),
                   opts.MarkPointItem(type_="average",name="平均值")
                                                         ]
                                                  )    
              )
    .add_yaxis("地下水资源总量",water_dixia,stack="stack1",)
    .add_yaxis("地表水资源总量",water_dibiao,stack="stack1",)
    .reversal_axis()
   
)
waterBar.render("水资源总量变化柱状图4.html")
```
