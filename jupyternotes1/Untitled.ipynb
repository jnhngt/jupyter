```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams["axes.unicode_minus"] = False        #显示负号
```

```python
#2023 : 43 85 62
#2024 : 54 87 67
#2025 : 45 91 65
```

```python
x = ['2023级','2024级','2025级']
y1 = [43,54,45]
y2 = [85,87,91]
y3 = [62,67,65]
plt .figure(figsize=(20,15))
```

```python
ax1 = plt.subplot(221)
ax1.plot(x,y1)
plt.title("大数据技术专业人数变化折线图")
plt.xlabel("年级",fontsize=8)
plt.ylabel("人数",fontsize=8)
ax2 = plt.subplot(222)
ax2.plot(x,y2)
plt.title("计算机应用专业人数变化折线图")
plt.xlabel("年级",fontsize=8)
plt.ylabel("人数",fontsize=8)
ax3 = plt.subplot(223)
ax3.plot(x,y3)
plt.title("物联网应用技术专业人数变化折线图")
plt.xlabel("年级",fontsize=8)
plt.ylabel("人数",fontsize=8)
ax4 = plt.subplot(224)
ax4.bar(x,y1)
plt.title("大数据技术专业人数变化柱状图")
plt.xlabel("年级",fontsize=8)
plt.ylabel("人数",fontsize=8)
plt.subplots_adjust(hspace=0.45)
plt.savefig("各专业人数变化子图.jpg")
plt .show()
```
