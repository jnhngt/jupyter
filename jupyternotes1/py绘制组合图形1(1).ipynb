```python
from pyecharts import options as opts
import pandas as pd
from pyecharts.charts import *
```

```python
df = pd.read_excel("消费量数据.xlsx")
df
```

```python
zb = df["指标"].values.tolist()
zb
```

```python
data_2024 = df["2024年"].values.tolist()
data_2024
```

```python
data_2023 = df["2023年"].values.tolist()
data_2023
```

```python
effect1 = (
    EffectScatter()
    
    .add_xaxis(zb)
    .add_yaxis("2024年",data_2024,symbol="diamond",symbol_size=12)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="2024年我国居民食品消费量"),
        legend_opts = opts.LegendOpts(pos_bottom="88%")
    )
)
```

```python
bar1 = (
    Bar()
    .add_xaxis(zb)
    .add_yaxis("2023年",data_2023)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="2023年我国居民食品消费量",pos_top="48%")
    )
)
```

```python
xftable = (
    Grid()
    .add(effect1,grid_opts=opts.GridOpts(pos_bottom = "55%"))
    .add(bar1,grid_opts=opts.GridOpts(pos_top = "55%"))
)
xftable.render("Grid组合图1.html")
```

```python
xftable = (
    Grid()
    .add(effect1,grid_opts=opts.GridOpts(pos_left = "55%"))
    .add(bar1,grid_opts=opts.GridOpts(pos_right = "55%"))
)
xftable.render("Grid组合图2.html")
```
