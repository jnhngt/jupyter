```python
from pyecharts.charts import *
from pyecharts import options as opts
import pandas as pd
```

```python
df = pd.read_excel("消费量数据.xlsx")
zb = df['指标'].values.tolist() 
data_2024 = df["2024年"].values.tolist() 
data_2023 = df["2023年"].values.tolist() 
```

```python
effect1 = (
    EffectScatter()
    .add_xaxis(zb)
    .add_yaxis("2024年", data_2024, symbol="diamond", symbol_size=12)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="2024年我国居民食品消费量图", pos_top="50%"),
        
    )
)
```

```python
bar1 = (
    Bar()
    .add_xaxis(zb)
    .add_yaxis("2023年", data_2023)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="2023年我国居民食品消费量图"),
        legend_opts=opts.LegendOpts(pos_bottom="88%")
    )
)
```

```python
xftable = (
    Grid()
    .add(effect1, grid_opts=opts.GridOpts(pos_top="55%"))
    .add(bar1, grid_opts=opts.GridOpts(pos_bottom="55%"))
)
```

```python
xftable.render("Grid组合图.html")
```

```python
xftable1 = (
    Grid()
    .add(effect1,grid_opts=opts.GridOpts(pos_left="55%"))
    .add(bar1,grid_opts=opts.GridOpts(pos_right="55%"))
)
```

```python
xftable1.render("Grid组合图1.html")
```

```python
pageimg = (
    Page()
    .add(effect1,bar1)
)
pageimg.render("Grid组合图2.html")
```

```python
tabimg = Tab()
tabimg.add(effect1,"")
tabimg.add(bar1,"")
tabimg.render("组合图4.html")
```
