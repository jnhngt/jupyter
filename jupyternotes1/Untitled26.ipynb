```python
from pyecharts import options as opts
from pyecharts.charts import Bar, Line, Page
```

```python
x_data = ["小米", "荣耀", "华为", "中兴", "魅族", "vivo", "OPPO"]
y_a = [107, 36, 102, 91, 51, 113, 45]
y_b = [104, 60, 33, 138, 105, 111, 91]
```

```python
bar = (
    Bar()
    .add_xaxis(x_data)
    .add_yaxis("商家A", y_a)
    .add_yaxis("商家B", y_b)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="组合图表-柱形图"),
        yaxis_opts=opts.AxisOpts(name="销售额(万元)", name_location="center", name_gap=30)
    )
)
```

```python
line = (
    Line()
    .add_xaxis(x_data)
    .add_yaxis("商家A", y_a)
    .add_yaxis("商家B", y_b)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="组合图表-折线图"),
        yaxis_opts=opts.AxisOpts(name="销售额(万元)", name_location="center", name_gap=30)
    )
)
```

```python
page = Page(page_title="顺序多图", interval=2)
page.add(bar, line)
page.render()
```
