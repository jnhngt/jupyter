```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams["axes.unicode_minus"] = False        #显示负号
import seaborn as sns
```

```python
sns.set_style('darkgrid')
sns.set_palette("pastel")
plt.rcParams['font.sans-serif'] = ['SimHei']
x = ['Q1','Q2','Q3','Q4']
jianzhi = [200,250,300,450]
piying = [100,120,150,220]
cixiu = [50,60,80,130]
plt.title("非遗产品四个季度的销售量堆积面积图")
plt.xlabel("季度")
plt.ylabel("销售量")
plt.stackplot(x,jianzhi,piying,cixiu,labels=["剪纸","皮影","刺绣"])
plt.legend()
plt.show()
```
