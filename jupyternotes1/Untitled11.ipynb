```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams["axes.unicode_minus"] = False        #显示负号
import wordcloud
import jieba
import imageio
```

```python
with open("demo.txt","r",encoding="utf-8") as f:
          text = f.read()
text
```

```python
text = " ".join(text)
text
```

```python
ls = jieba.lcut(text)
ls
```

```python
text = " ".join(ls)
text
```

```python
mk = imageio.imread_v2("ac8138d162953c70.webp")
wc = wordcloud.WordCloud(
    mask = mk,
    background_color='white',
    width=2000,
    height=1500,
    stopwords=['和','的'],
    font_path="simhei.ttf")
wc.generate(text)
plt.imshow(wc)
plt.axis("off")
wc.to_file("中华人民共和国国家发展规划法1.jpg")
plt.show()
```
