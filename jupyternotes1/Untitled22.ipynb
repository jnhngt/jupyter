```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']  
plt.rcParams["axes.unicode_minus"] = False    
```

```python
food_categories = ['谷物及薯类', '肉禽蛋奶类', '蔬菜类', '水果类', '水产品类', '其他食品']
expense_ratios = [25.3, 32.1, 18.6, 10.8, 9.2, 4.0]  
adjusted_ratios = [x + 0.5 for x in expense_ratios]
```

```python
new_categories = food_categories[:-1]  
new_ratios = adjusted_ratios[:-1]   
```

```python
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']
plt.figure(figsize=(8, 8))
plt.pie(new_ratios,
        labels=new_categories,
        autopct='%1.1f%%',
        colors=colors,
        shadow=True)
plt.axis('equal')
plt.title('2024年全国居民核心食品消费支出结构',
          fontsize=14,
          fontweight='bold',
          color='#36454F')
plt.legend(loc='center left', bbox_to_anchor=(1, 0.5), fontsize=8)
plt.savefig('food_consumption_pie_105.png', dpi=300, bbox_inches='tight')
plt.show()
```
