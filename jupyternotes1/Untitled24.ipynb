```python
from pyecharts.charts import EffectScatter
from pyecharts import options as opts
import pandas as pd
```

```python
df = pd.read_excel("消费量数据.xlsx")
```

```python
zb = df['指标'].values.tolist() 
```

```python
data_2024 = df["2024年"].values.tolist() 
```

```python
data_2023 = df["2023年"].values.tolist() 
```

```python
effect1 = (
    EffectScatter()
    .add_xaxis(zb)
    .add_yaxis("",data_2024,symbol='diamond',symbol_size=12)
    .add_yaxis("",data_2023,symbol='pin',symbol_size=12)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="人口总量变化柱状图",subtitle="2016年-2025年"),
        
    )
)

effect1.render("2024我国居民消费量涟漪散点图.html")
```
