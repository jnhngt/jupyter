```python
from pyecharts import options as opts
import pandas as pd
from pyecharts.charts import *#Tab
```

```python
df = pd.read_excel("消费量数据.xlsx")
df
```

```python
zb = df["指标"].values.tolist()
zb
```

```python
data_2024 = df["2024年"].values.tolist()
data_2024
```

```python
data_2023 = df["2023年"].values.tolist()
data_2023
```

```python
effect1 = (
    EffectScatter()
    
    .add_xaxis(zb)
    .add_yaxis("2024年",data_2024,symbol="diamond",symbol_size=12)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="2024年我国居民食品消费量")
    )
)
bar1 = (
    Bar()
    .add_xaxis(zb)
    .add_yaxis("2023年",data_2023)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="2023年我国居民食品消费量")
    )
)
```

```python
tabimg =Tab()
tabimg.add(effect1,"2024年居民消费量涟漪散点图")
tabimg.add(bar1,"2023年居民消费量柱状图")
tabimg.render("组合图4.html")
```
