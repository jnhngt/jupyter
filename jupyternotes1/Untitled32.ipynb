```python
from pyecharts import options as opts
from pyecharts.charts import  Bar, Line, Grid, Page, Tab
```

```python
months = ["1月", "2月", "3月", "4月", "5月", "6月"]
sales = [120, 150, 180, 220, 200, 250]      
orders = [8.5, 10.2, 11.8, 14.5, 13.8, 16.2]  
users = [25, 32, 40, 45, 38, 50]              
repurchase = [28, 30, 35, 38, 36, 40] 
```

```python
bar = (
    Bar()
    .add_xaxis(months)
    .add_yaxis("订单量(万单)", orders)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="上下组合图表"),
        yaxis_opts=opts.AxisOpts(name="订单量(万单)"),
        legend_opts=opts.LegendOpts(pos_right="5%",pos_top="5%")
    )
)
```

```python
line = (
    Line()
    .add_xaxis(months)
    .add_yaxis("销售额(万元)", sales)
    .set_global_opts(
        yaxis_opts=opts.AxisOpts(name="销售额(万元)"),
        legend_opts=opts.LegendOpts(pos_right="5%",pos_bottom="5%")
    )
)
```

```python
yftable = (
    Grid()
    .add(bar,grid_opts=opts.GridOpts(pos_bottom = "55%"))
    .add(line,grid_opts=opts.GridOpts(pos_top = "55%"))
)
yftable.render("Grid组合图2.html")
```
