```python
from pyecharts.charts import Line
from pyecharts import options as opts
import pandas as pd
```

```python
df = pd.read_excel("水资源数据.xlsx")
```

```python
zong_data = df.iloc[0,1:]
zong_data
```

```python
dibiao_data = df.iloc[1,1:]
dibiao_data
```

```python
dixia_data = df.iloc[2,1:]
dixia_data
```

```python
year = zong_data.index.to_list()
year.reverse()
year
```

```python
zong_data.values
```

```python
water_zong = zong_data.values.tolist()[::-1]
```

```python
waterline = (
    Line()
    .add_xaxis(year)
    .add_yaxis("水资源总量",water_zong,
               symbol='pin',
               symbol_size=12,
               areastyle_opts = opts.AreaStyleOpts(color="#0E84F2",opacity="0.5"),
               markpoint_opts = opts.MarkPointOpts(data =[
                   opts.MarkPointItem(type_="max",name="最大值"),
                   opts.MarkPointItem(type_="min",name="最小值"),
                   opts.MarkPointItem(type_="average",name="平均值")
                                                         ]
                                                  )    
              )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="水资源总量变化折线图",subtitle="2016年-2024年"),
        yaxis_opts=opts.AxisOpts(name="总量(亿立方米)"),
        xaxis_opts=opts.AxisOpts(name="年份")
    )
)
waterline.render("水资源总量变化折线图.html")
```

```python
waterline = (
    Line()
    .add_xaxis(year)
    .add_yaxis("水资源总量",water_zong,
               symbol='pin',
               symbol_size=12,
               areastyle_opts = opts.AreaStyleOpts(color="#0E84F2",opacity="0.5"),
               markline_opts = opts.MarkLineOpts(data =[
                   opts.MarkLineItem(type_="max",name="最大值"),
                   opts.MarkLineItem(type_="min",name="最小值"),
                   opts.MarkLineItem(type_="average",name="平均值")
                                                         ]
                                                  )    
              )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="水资源总量变化折线图",subtitle="2016年-2024年"),
        yaxis_opts=opts.AxisOpts(name="总量(亿立方米)"),
        xaxis_opts=opts.AxisOpts(name="年份")
    )
)
waterline.render("水资源总量变化折线图2.html")
```

```python
waterline = (
    Line()
    .add_xaxis(year)
    .add_yaxis("水资源总量",water_zong)
    .add_yaxis("地表水资源总量", dibiao_data)
    .add_yaxis("地下水资源总量",dixia_data)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="各种水资源总量变化折线图",subtitle="2016年-2024年"),
        yaxis_opts=opts.AxisOpts(name="总量(亿立方米)",name_location="center"),
        xaxis_opts=opts.AxisOpts(name="年份",name_location="center"),
        legend_opts=opts.LegendOpts(pos_top="8%",pos_left="55%"),      
        toolbox_opts=opts.ToolboxOpts(is_show=True)
    )
)
waterline.render("各种水资源总量变化折线图.html")
```
