```python
import pandas as pd
import seaborn as sns
from pyecharts import options as opts
from pyecharts.charts import *
```

```python
total = pd.read_excel('总人口.xlsx')
csz = pd.read_excel('人口出生率、死亡率和自然增长率.xlsx')
age = pd.read_excel('人口年龄结构和抚养比.xlsx')
average = pd.read_excel('人口平均预期寿命.xlsx')
census = pd.read_excel('人口普查人口基本情况.xlsx') 
total
```

```python
years = total.columns[1:].tolist()[::-1]  # 获取年份列表，并反转顺序
```

```python
people = total.iloc[0, 1:].values.tolist()[::-1]
people = [int(i) for i in people]
```

```python
men = total.iloc[1, 1:].values.tolist()[::-1]
men = [int(i) for i in men]
```

```python
women = total.iloc[2, 1:].values.tolist()[::-1]
women = [int(i) for i in women]
```

```python
city = total.iloc[3, 1:].values.tolist()[::-1]
city = [int(i) for i in city]
```

```python
rural = total.iloc[4, 1:].values.tolist()[::-1]
rural = [int(i) for i in rural]
```

```python
die = csz.iloc[1,1:].values.tolist()[::-1]
die
```

```python
die2025 = csz.iloc[1,1:2].values.tolist()[::-1]
```

```python
new = csz.iloc[0,1:].values.tolist()[::-1]
new
```

```python
new2025 = csz.iloc[0,1:2].values.tolist()[::-1]
```

```python
growth = csz.iloc[2,1:].values.tolist()[::-1]
```

```python
bar = (
    Bar()
    .add_xaxis(years)
    .add_yaxis("男性人口",men)
    .add_yaxis("女性人口",women)
    .set_global_opts(
        yaxis_opts=opts.AxisOpts(name="人数(万人)"),
        legend_opts=opts.LegendOpts(pos_right="5%",pos_bottom="5%")
    )
)
```

```python
bar.render("bar.html")
```

```python
line = (
    Line(init_opts=opts.InitOpts(width="1000px", height="420px"))
    .add_xaxis(years)
    .add_yaxis("年末总人口 (万人)", people)
    .set_global_opts(
        yaxis_opts=opts.AxisOpts(name="人数(万人)"),
        legend_opts=opts.LegendOpts(pos_right="5%",pos_bottom="5%")
    )
)
```

```python
line.render("line.html")
```

```python
pie = (
    Pie()
    .add("", [["2025死亡率",die2025],["2025出生率",new2025]])
    .set_global_opts(title_opts=opts.TitleOpts(title="2025年人口出生率、死亡率和自然增长率"))
    .set_series_opts(label_opts=opts.LabelOpts(formatter="{b}: {c}"))
)
```

```python
pie.render("pie.html")
```

```python
line2 = (
    Line()
    .add_xaxis(years)
    .add_yaxis("人口自然增长率%", growth)
    .add_yaxis("出生率%", new)
    .add_yaxis("死亡率%", die)
    .set_series_opts(
        markline_opts=opts.MarkLineOpts(
            data=[
                opts.MarkLineItem(
                    name="",
                    y=0
                )
            ],
            linestyle_opts=opts.LineStyleOpts(
                width=2, 
                color='#FF0000'
            ),
            symbol="none"
        )
    )
)
```

```python
line2.render("line2.html")
```

```python
bar1 = (
    Bar()
    .add_xaxis(years)
    .add_yaxis("",city,stack='stack1')
    .add_yaxis("",rural,stack='stack1')
)
```

```python
bar1.render("bar1.html")
```

```python
page1  = (

    Page()
    .add(bar,line,pie,line2,bar1)
    
)

page1.render("组合图3.html")
```

```python
page1.render("page1.html")
```
