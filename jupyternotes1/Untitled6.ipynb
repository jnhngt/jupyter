```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_excel("5-1.xlsx")
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.boxplot(y="技术实现成绩",data=df,color='brown')
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.boxplot(y="创新思维成绩",data=df)
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.boxplot(y="工程规范成绩",data=df)
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.boxplot(y="项目开发耗时",data=df,color='yellow',fliersize=10,flierprops=dict(marker='*',markerfacecolor='red'))
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.boxplot(y="项目开发耗时",hue="性别",data=df,fliersize=10,flierprops=dict(marker='*',markerfacecolor='red'))
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.boxplot(y="项目开发耗时",hue="编程基础等级",data=df,fliersize=10,flierprops=dict(marker='*',markerfacecolor='red'))
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.boxplot(y="技术实现成绩",palette="viridis",hue="编程基础等级",hue_order=['初级','中级','高级'],data=df,fliersize=10,flierprops=dict(marker='*',markerfacecolor='red'))
```
