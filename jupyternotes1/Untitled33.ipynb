```python
from pyecharts import options as opts
from pyecharts.charts import  Bar, Line, Grid, Page, Tab
```

```python
months = ["1月", "2月", "3月", "4月", "5月", "6月"]
sales = [120, 150, 180, 220, 200, 250]      
orders = [8.5, 10.2, 11.8, 14.5, 13.8, 16.2]  
users = [25, 32, 40, 45, 38, 50]              
repurchase = [28, 30, 35, 38, 36, 40] 
```

```python
chart1 = (
    Bar()
    .add_xaxis(months)
    .add_yaxis("销售额(万元)", sales)
    .set_global_opts(title_opts=opts.TitleOpts(title="销售额柱状图"))
)

chart2 = (
    Line()
    .add_xaxis(months)
    .add_yaxis("订单量(万单)", orders)
    .set_global_opts(title_opts=opts.TitleOpts(title="订单量折线图"))
)

chart3 = (
    Line()
    .add_xaxis(months)
    .add_yaxis("复购率(%)", repurchase)
    .set_global_opts(title_opts=opts.TitleOpts(title="复购率趋势图"))
)

chart4 = (
    Bar()
    .add_xaxis(months)
    .add_yaxis("新用户(千人)",users)
    .set_global_opts(title_opts=opts.TitleOpts(title="月度新用户趋势图"))
)

tab = Tab()
tab.add(chart1, "销售额柱状图")
tab.add(chart2, "订单量折线图")
tab.add(chart3, "复购率趋势图")
tab.render("选项卡Tab图表.html")
```

```python
page = Page(page_title="test")
page.add(chart1)
page.render("销售额柱状图.html")
```

```python
page = Page(page_title="test")
page.add(chart2)
page.render("订单量折线图.html")
```
