```python
from pyecharts.charts import Funnel
from pyecharts import options as opts
import pandas as pd
```

```python
df = pd.read_excel("消费量数据.xlsx")
```

```python
zb = df['指标']
```

```python
data_2024 = df["2024年"]
data_2024
```

```python
Funnel1 = (

    Funnel()
    .add("2024年我国居民食品消费量",
         [list(z) for z in zip(zb,data_2024)],
         label_opts= opts.LabelOpts(position="inside",formatter="{b}:{c}")
        )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="2024年我国居民食品消费量")
    )
)
Funnel1.render("2024年我国居民食品消费量漏斗图.html")
```
