```python
from pyecharts.charts import Bar
from pyecharts.charts import Line
from pyecharts import options as opts
import pandas as pd
```

```python
df = pd.read_excel("总人口数据.xlsx")
```

```python
peoplezong_data = df.iloc[0,1:]
peoplezong_data
```

```python
year = peoplezong_data.index.to_list()[::-1]
year
```

```python
people_zong = peoplezong_data.values.tolist()[::-1]
people_zong1 = [int(i) for i in peoplezong_data]
people_zong
```

```python
peopleBar = (
    Bar()
    .add_xaxis(year)
    .add_yaxis("人口总量",people_zong1,
               markpoint_opts = opts.MarkPointOpts(data =[
                   opts.MarkPointItem(type_="max",name="最大值"),
                   opts.MarkPointItem(type_="min",name="最小值"),
                   opts.MarkPointItem(type_="average",name="平均值")
                                                         ]
                                                  )    
              )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="人口总量变化柱状图",subtitle="2016年-2025年"),
        yaxis_opts=opts.AxisOpts(name="总量(万人)"),
        xaxis_opts=opts.AxisOpts(name="年份")
    )
)
peopleBar.render("人口总量变化柱状图.html")
```

```python
peopleLine = (
    Line()
    .add_xaxis(year)
    .add_yaxis("人口总量",people_zong1,
               markpoint_opts = opts.MarkPointOpts(data =[
                   opts.MarkPointItem(type_="max",name="最大值"),
                   opts.MarkPointItem(type_="min",name="最小值"),
                   opts.MarkPointItem(type_="average",name="平均值")
                                                         ]
                                                  )    
              )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="人口总量变化折线图",subtitle="2016年-2025年"),
        yaxis_opts=opts.AxisOpts(name="总量(万人)"),
        xaxis_opts=opts.AxisOpts(name="年份")
    )
)
peopleLine.render("人口总量变化折线图.html")
```
