```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_excel("身高数据.xlsx", sheet_name="Sheet1")

# 2）提取"性别"和"身高(cm)"列
height_data = df[["性别", "身高 (cm)"]].copy()

print("数据预览：")
print(height_data.head())
print(f"\n总人数：{len(height_data)}")
print(f"男生人数：{len(height_data[height_data['性别']=='男'])}")
print(f"女生人数：{len(height_data[height_data['性别']=='女'])}")
```

```python
sns.set_style("darkgrid")                                   
plt.rcParams['font.sans-serif'] = ['SimHei']                
plt.rcParams['axes.unicode_minus'] = False                   
plt.figure(figsize=(10, 7))

sns.boxplot(
    data=height_data,
    y="身高 (cm)",
    color="#4CAF50",  # 指定颜色
    width=0.4
)

median_all = height_data["身高 (cm)"].median()
plt.text(0, median_all + 0.5, f"中位数：{median_all:.0f}cm",
         ha='center', va='bottom', fontsize=11, fontweight='bold',
         color='#333333')

plt.title("大数据技术专业班级整体身高分布箱型图", fontsize=14, fontweight='bold')
plt.ylabel("身高 (cm)", fontsize=12)
plt.xticks([])
plt.tight_layout()
plt.savefig("整体身高箱型图_张宇洋.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10, 7))

sns.boxplot(
    data=height_data,
    x="性别",
    y="身高 (cm)",
    hue="性别",
    palette="Set2", 
    width=0.4,
    legend=False
)

median_male = height_data[height_data["性别"]=="男"]["身高 (cm)"].median()
median_female = height_data[height_data["性别"]=="女"]["身高 (cm)"].median()
plt.text(0, median_female + 0.5, f"中位数：{median_female:.0f}cm",
         ha='center', va='bottom', fontsize=11, fontweight='bold', color='#d62728')
plt.text(1, median_male + 0.5, f"中位数：{median_male:.0f}cm",
         ha='center', va='bottom', fontsize=11, fontweight='bold', color='#1f77b4')

plt.title("大数据技术专业班级男女生身高分布箱型图", fontsize=14, fontweight='bold')
plt.xlabel("性别", fontsize=12)
plt.ylabel("身高 (cm)", fontsize=12)
plt.tight_layout()
plt.savefig("bigdata_class_height_boxplot_张宇洋.png", dpi=300, bbox_inches='tight')
plt.show()
```
