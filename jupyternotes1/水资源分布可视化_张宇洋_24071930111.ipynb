```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# 从Excel读取数据（不写死数据）
df = pd.read_excel("年度数据_国家统计局.xlsx", header=2)
df = df.set_index('指标')
```

```python
# 提取数据
water_total = df.loc['水资源总量 (亿立方米) '].values.tolist()
surface_water = df.loc['地表水资源量 (亿立方米) '].values.tolist()
ground_water = df.loc['地下水资源量 (亿立方米) '].values.tolist()
per_capita = df.loc['人均水资源量 (立方米/人) '].values.tolist()

# 年份标签
year_label = df.columns.tolist()

# 转换为数值型
water_total = [float(x) for x in water_total]
surface_water = [float(x) for x in surface_water]
ground_water = [float(x) for x in ground_water]
per_capita = [float(x) for x in per_capita]
```

```python
# 描述性统计
avg = sum(water_total)/len(water_total)
max_val = max(water_total)
max_yr = year_label[water_total.index(max_val)]
min_val = min(water_total)
min_yr = year_label[water_total.index(min_val)]
print(f"数据来源：国家统计局年度数据导出Excel")
print(f"年份：{'、'.join(year_label)}")
print(f"多年平均水资源总量：{avg:.1f}亿立方米")
print(f"最高：{max_yr}（{max_val:.1f}亿立方米）")
print(f"最低：{min_yr}（{min_val:.1f}亿立方米）")

x_pos = range(len(year_label))
```

## 图1：Matplotlib基础柱状图

```python
plt.figure(figsize=(12,7))
plt.bar(x_pos,water_total,width=0.6,color='steelblue',edgecolor='black',alpha=0.85)
for i in x_pos:
    plt.text(i,water_total[i]+200,f'{water_total[i]:.0f}',ha='center',fontsize=10)
plt.axhline(y=avg,color='red',linestyle='--',linewidth=1.5,label=f'均值：{avg:.1f}亿m³')
plt.title("我国水资源总量变化趋势——Matplotlib基础柱状图",fontsize=14)
plt.xlabel("年份",fontsize=12)
plt.ylabel("水资源总量（亿立方米）",fontsize=12)
plt.xticks(x_pos,year_label,fontsize=10)
plt.legend()
plt.grid(axis='y',linestyle=':',alpha=0.6)
plt.tight_layout()
plt.savefig('图1_水资源基础柱状图_张宇洋_24071930111.png',dpi=300,bbox_inches='tight')
plt.show()
```

## 图2：Seaborn调色板优化柱状图

```python
sns.set_style("darkgrid")                                      #必须在显示中文前
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(12,7))
colors = sns.color_palette('husl',n_colors=len(year_label))
plt.bar(x_pos,water_total,width=0.6,color=colors,edgecolor='gray',alpha=0.9)
for i in x_pos:
    plt.text(i,water_total[i]+150,f'{water_total[i]:.0f}',ha='center',fontsize=10,fontweight='bold')
plt.axhline(y=avg,color='crimson',linestyle='--',linewidth=2,alpha=0.8,label=f'均值：{avg:.1f}亿m³')

max_idx = water_total.index(max_val)
min_idx = water_total.index(min_val)
plt.annotate(f'最高：{max_yr} {max_val:.0f}亿m³',xy=(max_idx,max_val),
             xytext=(max_idx+0.3,max_val+1500),
             arrowprops=dict(arrowstyle='->',color='darkgreen',lw=1.5),
             fontsize=10,color='darkgreen',fontweight='bold')
plt.annotate(f'最低：{min_yr} {min_val:.0f}亿m³',xy=(min_idx,min_val),
             xytext=(min_idx-0.3,min_val-2000),
             arrowprops=dict(arrowstyle='->',color='darkred',lw=1.5),
             fontsize=10,color='darkred',fontweight='bold')

plt.title("我国水资源总量变化趋势——Seaborn调色板优化",fontsize=14,fontweight='bold')
plt.xlabel("年份",fontsize=12)
plt.ylabel("水资源总量（亿立方米）",fontsize=12)
plt.xticks(x_pos,year_label,fontsize=10)
plt.legend()
plt.tight_layout()
plt.savefig('图2_水资源Seaborn优化柱状图_张宇洋_24071930111.png',dpi=300,bbox_inches='tight')
plt.show()
```

## 图3：水资源构成分组对比图

```python
sns.set_style("darkgrid")                                      #必须在显示中文前
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(12,7))
w = 0.2
palette = sns.color_palette('viridis',3)
plt.bar([i-w for i in x_pos],water_total,width=w,color=palette[2],label='水资源总量',alpha=0.9)
plt.bar(x_pos,surface_water,width=w,color=palette[0],label='地表水资源量',alpha=0.9)
plt.bar([i+w for i in x_pos],ground_water,width=w,color=palette[1],label='地下水资源量',alpha=0.9)
plt.title("我国水资源构成对比图——Seaborn Viridis调色板",fontsize=14,fontweight='bold')
plt.xlabel("年份",fontsize=12)
plt.ylabel("水资源量（亿立方米）",fontsize=12)
plt.xticks(x_pos,year_label,fontsize=10)
plt.legend(fontsize=10)
plt.grid(axis='y',linestyle=':',alpha=0.4)
plt.tight_layout()
plt.savefig('图3_水资源构成对比图_张宇洋_24071930111.png',dpi=300,bbox_inches='tight')
plt.show()
```
