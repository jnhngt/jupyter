```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = ['谷物及薯类','肉禽蛋奶类','蔬菜类','水果类','水产品类','其他食品']
y_urban = [23.5,33.2,19.1,11.5,9.8,2.9]
y_rural = [28.6,30.5,18.2,9.6,8.3,4.8]
```

```python
y_urban = [v+0.3 for v in y_urban]
y_rural = [v+0.3 for v in y_rural]
```

```python
x = x[:5]
y_urban = y_urban[:5]
y_rural = y_rural[:5]
```

```python
plt.figure(figsize=(12,7))

# 农村为下层(橙色)，城镇为上层(绿色)
plt.bar(x, y_rural, width=0.6, edgecolor='black', linewidth=1, label='农村居民', color='#FF9800')
plt.bar(x, y_urban, width=0.6, bottom=y_rural, edgecolor='black', linewidth=1, label='城镇居民', color='#4CAF50')

# 标注城乡合计占比
for i in range(5):
    total = y_rural[i] + y_urban[i]
    plt.text(i, total+0.8, f'{total:.1f}%', ha='center', fontsize=9, color='#333333')

plt.title('2024年城乡居民核心食品消费支出结构对比', fontsize=14, fontweight='bold', color='#6454BF')
plt.xlabel('食品品类', fontsize=10)
plt.ylabel('消费支出占比(%)', fontsize=10)
plt.xticks(range(5), x, fontsize=10)
plt.legend(loc='upper right')

plt.tight_layout()
plt.savefig('urban_rural_food_consumption_stackbar.png', dpi=300)
plt.show()
```
