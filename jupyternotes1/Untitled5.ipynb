```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_excel("5-1.xlsx")
```

```python
sns.set_style("darkgrid")                                      #必须在显示中文前
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.countplot(x="性别",data=df,palette='deep')
plt.ylabel("人数")
plt.show()
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.countplot(x="年级",data=df,order=['大二','大三'],palette='inferno')
plt.ylabel("人数")
plt.show()
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.countplot(x="编程基础等级",data=df,order=['初级','中级','高级'],palette='magma')
plt.ylabel("人数")
plt.show()
```

```python
sns.set_style("darkgrid")                                      #必须在显示中文前
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.countplot(x="性别",data=df,palette='deep',hue='年级')
plt.ylabel("人数")
plt.show()
```

```python
sns.set_style("darkgrid")                                      #必须在显示中文前
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.countplot(x="性别",data=df,palette='deep',hue='编程基础等级',)
plt.ylabel("人数")
plt.legend(bbox_to_anchor=(1.02,0.96))
plt.show()
```
