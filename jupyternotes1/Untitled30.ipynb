```python
from pyecharts import options as opts
from pyecharts.charts import Page,Bar
```

```python
quarter = ['Q1', 'Q2', 'Q3', 'Q4']
Product_A = [85, 92, 88, 95]
Product_B = [85, 92, 88, 95]
Product_C = [85, 92, 88, 95]
```

```python
bar_1 = (
    Bar()
    .add_xaxis(quarter)
    .add_yaxis("产品A",Product_A)
    .add_yaxis("产品B",Product_B)
    .add_yaxis("产品C",Product_C)
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="组合图表-柱形图",
            subtitle="单位:万元" ),
        yaxis_opts=opts.AxisOpts(name="销售额(万元)", name_location="center", name_gap=30)
    )
)
```

```python
page = Page(page_title="test2_1")
page.add(bar_1)
page.render("product_sales_bar_1.html")
```

```python
bar_2 = (
    Bar()
    .add_xaxis(quarter)
    .add_yaxis("产品A",Product_A, stack="stack1")
    .add_yaxis("产品B",Product_B, stack="stack1")
    .add_yaxis("产品C",Product_C, stack="stack1")
    .reversal_axis()
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="组合图表-柱形图",
            subtitle="单位:万元" ),
        yaxis_opts=opts.AxisOpts(name="销售额(万元)", name_location="center", name_gap=30)
    )
)
```

```python
page = Page(page_title="test2_2")
page.add(bar_2)
page.render("product_sales_bar_2.html")
```
