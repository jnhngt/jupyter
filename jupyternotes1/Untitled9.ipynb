```python
import pandas as pd
import numpy as np
import sklearn
from sklearn.tree import DecisionTreeClassifier
```

```python
data = {
    '客户':['客户1','客户2','客户3'],
    '年龄':[65,35,24],
    '性别':['0','1','0'],
    '是否点甜品':['0','1','0']
}
df = pd.DataFrame(data)
```

```python
df
```

```python
test_data = {
    '客户':['客户4'],
    '年龄':['33'],
    '性别':['0']
}
test_df = pd.DataFrame(test_data)
```

```python
test_df
```

```python
model = DecisionTreeClassifier(random_state=0)
```

```python
model = model.fit(df[['年龄','性别']],df['是否点甜品'])
```

```python
prediction = model.predict(test_df[['年龄','性别']])
```

```python
result = '是' if prediction[0]==1 else '否'
```

```python
print(f'客户4点甜品吗?:{result}')
```
