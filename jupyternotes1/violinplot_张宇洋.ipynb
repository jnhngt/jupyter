```python
# 导入所需库
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 第一步：数据预处理
# 1）读取Excel数据表
df = pd.read_excel("身高数据.xlsx", sheet_name="Sheet1")

# 2）提取"性别"和"身高(cm)"列
height_data = df[["性别", "身高 (cm)"]].copy()

print("数据预览：")
print(height_data.head())
print(f"\n总人数：{len(height_data)}")
print(f"男生人数：{len(height_data[height_data['性别']=='男'])}")
print(f"女生人数：{len(height_data[height_data['性别']=='女'])}")
```

```python
# 基础样式设置
sns.set_style("darkgrid")                                      # 设置Seaborn绘图风格
plt.rcParams['font.sans-serif'] = ['SimHei']                  # 解决中文乱码
plt.rcParams['axes.unicode_minus'] = False                    # 正常显示负号

# 第二步：绘制整体身高分布小提琴图
plt.figure(figsize=(10, 7))

sns.violinplot(
    data=height_data,
    y="身高 (cm)",
    color="#4CAF50",
    width=0.5
)

# 标注中位数
median_all = height_data["身高 (cm)"].median()
plt.text(0, median_all + 0.5, f"中位数：{median_all:.0f}cm",
         ha='center', va='bottom', fontsize=11, fontweight='bold', color='#333333')

plt.title("大数据技术专业班级整体身高分布小提琴图", fontsize=14, fontweight='bold')
plt.ylabel("身高 (cm)", fontsize=12)
plt.xticks([])
plt.tight_layout()
plt.savefig("整体身高小提琴图_张宇洋.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 第三步：绘制按性别分组的身高分布小提琴图
plt.figure(figsize=(10, 7))

sns.violinplot(
    data=height_data,
    x="性别",
    y="身高 (cm)",
    hue="性别",
    palette="Set2",
    width=0.5,
    legend=False
)

# 分别标注男女身高中位数
median_male = height_data[height_data["性别"]=="男"]["身高 (cm)"].median()
median_female = height_data[height_data["性别"]=="女"]["身高 (cm)"].median()
plt.text(0, median_female + 0.5, f"中位数：{median_female:.0f}cm",
         ha='center', va='bottom', fontsize=11, fontweight='bold', color='#d62728')
plt.text(1, median_male + 0.5, f"中位数：{median_male:.0f}cm",
         ha='center', va='bottom', fontsize=11, fontweight='bold', color='#1f77b4')

plt.title("大数据技术专业班级男女生身高分布小提琴图", fontsize=14, fontweight='bold')
plt.xlabel("性别", fontsize=12)
plt.ylabel("身高 (cm)", fontsize=12)
plt.tight_layout()
plt.savefig("bigdata_class_height_violinplot_张宇洋.png", dpi=300, bbox_inches='tight')
plt.show()

print("图表已保存：")
print("  1. 整体身高小提琴图_张宇洋.png")
print("  2. bigdata_class_height_violinplot_张宇洋.png")
```

```python
# 数据分析小结
print("数据分析小结：")
print(f"  整体中位数：{median_all:.0f} cm")
print(f"  男生中位数：{median_male:.0f} cm")
print(f"  女生中位数：{median_female:.0f} cm")
print(f"  男女生身高差距：{median_male - median_female:.0f} cm")
```
