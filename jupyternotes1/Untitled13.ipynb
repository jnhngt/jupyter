```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams["axes.unicode_minus"] = False        #显示负号
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
x = np.array([1,1.5,2,2.5,3,3.5,4,4.5,5,5.5,6]).reshape(-1,1)
```

```python
y = np.array([1,1.9,2.8,3.9,4.8,5.9,6.8,7.9,8.8,9.9,10.8])
```

```python
model = LinearRegression()
model.fit(x,y)
```

```python
y_pred = model.predict(x)
```

```python
mse = mean_squared_error(y,y_pred)
```

```python
print(f"均方误差:{mse}")
```

```python
w = model.coef_[0]
```

```python
b = model.intercept_
```

```python
print(f"斜率 w:{w}")
print(f"截距 b:{b}")
```

```python
plt.scatter(x,y,color='blue')
plt.plot(x,y,color='red',linewidth=2)
plt.xlabel('房屋面积')
plt.ylabel('房屋价格')
plt.title('一元线性回归拟合线')
plt.show()
```
