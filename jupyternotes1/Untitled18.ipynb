```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
```

```python
df = pd.read_excel("test1.xlsx")
```

```python
peoplezong_data = df.iloc[0,1:]
peoplezong_data
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
plt.figure(figsize=(14,8))
plt.bar(df['省份'], df['实名志愿者数(万人)'],width=(0.35),color='#4287f5',alpha=0.8)
plt.title('各省实名志愿者数')
plt.xticks(fontsize=10, rotation=30)
plt.yticks(fontsize=10)
plt.xlabel('省份',fontsize=12)
plt.ylabel('志愿者数(万人)',fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(14,8))
plt.bar(df['省份'], df['年度服务时长(万小时)'],width=(0.35),color='#4287f5',alpha=0.8)
plt.title('各省实名志愿者数')
plt.xticks(fontsize=10, rotation=30)
plt.yticks(fontsize=10)
plt.xlabel('省份',fontsize=10)
plt.ylabel('志愿者数(万人)',fontsize=10)
plt.show()
```

```python
hours_qwh = df['年度服务时长(万小时)'] / 1000
hours_qwh
```

```python
wanren_qwh = df['实名志愿者数(万人)'] / hours_qwh
```

```python
plt.figure(figsize=(14,8))
plt.bar(df['省份'],wanren_qwh,width=(0.35), color='#42f5b3', alpha=0.8)
plt.title('2023年全国8省份志愿服务核心数据对比')
plt.xticks(fontsize=10, rotation=30)
plt.yticks(fontsize=10)
plt.xlabel('省份',fontsize=10)
plt.ylabel('数量(万人/千万小时)',fontsize=10)
sns.set_style('darkgrid')
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
plt.savefig('provincial_volunteer_analysis.png', dpi=200)
plt.show()
```
