```python
from pyecharts import options as opts
from pyecharts.charts import Line, Page
```

```python
years = ["2018", "2019", "2020","2021", "2022", "2023"]
national_level = [49,51,53, 57,60, 63]
autonomous_level = [337,359,382,405,428,451] 
league_level=[1126,1189,1253,1317,1381,1445] 
```

```python
line = (
    Line()
    .add_xaxis(years)
    .add_yaxis("国家级非遗项目数量", national_level)
    .add_yaxis("自治区级非遗项目数量", autonomous_level)
    .add_yaxis("盟市级非遗项目数量", league_level)
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="2018-2023年内蒙古自治区非遗项目数量变化趋势",
            subtitle="内蒙古自治区文化和旅游厅"                   
        ),
        xaxis_opts=opts.AxisOpts(name="「年份」", name_location="center", name_gap=30),
        yaxis_opts=opts.AxisOpts(name="「非遗项目数量(项)」", name_location="center", name_gap=30)
    )
)
```

```python
page = Page(page_title="test1")
page.add(line)
page.render("nmg_heritage_trend.html")
```
