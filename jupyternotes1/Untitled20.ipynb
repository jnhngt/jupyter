```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
```

```python
years = ['2016','2017','2018','2019','2020','2021','2022','2023','2024']
volenteer_num = [830,1080,1370,1720,2190,2180,2250,2320,2380]
service_hours = [8.5,11.2,15.8,20.5,28.3,32.7,36.1,41.8,53.6]
```

```python
fig, ax1 = plt.subplots(figsize=(12, 7))

line1, = ax1.plot(years,volenteer_num, 'o-', color='#1E90FF', linewidth=3, markersize=8, label='全国实名志愿者数（万人）')
ax1.set_xlabel('年份', fontsize=12)
ax1.set_ylabel('全国实名志愿者数（万人）', color='#1E90FF', fontsize=12)
ax1.tick_params(axis='y', labelcolor='#1E90FF')
ax1.set_xticks(years)
ax1.tick_params(axis='x', labelsize=10)

ax2 = ax1.twinx()
line2, = ax2.plot(years,service_hours, 's-', color='#FF6347', linewidth=3, markersize=8, label='累计志愿服务时长（亿小时）')
ax2.set_ylabel('累计志愿服务时长（亿小时）', color='#FF6347', fontsize=12)
ax2.tick_params(axis='y', labelcolor='#FF6347')

for x, y in zip(years,volenteer_num):
    ax1.text(x, y + 20, str(y), ha='center', fontsize=10)
for x, y in zip(years,service_hours):
    ax2.text(x, y + 1, str(y), ha='center', fontsize=10)

plt.title('2016-2024年全国志愿服务核心数据趋势', fontsize=18, fontweight='bold')
ax1.grid(axis='y', linestyle='--', alpha=0.7, color='gray')

plt.legend([line1, line2], [line1.get_label(), line2.get_label()], loc='upper left', bbox_to_anchor=(1.02, 1))

plt.tight_layout()
plt.savefig('volunteer_service_trend.png', dpi=300, bbox_inches='tight')
plt.show()
```
